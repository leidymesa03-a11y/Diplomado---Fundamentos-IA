## 1. Investigar es responder una pregunta con evidencia

Una investigación no comienza necesariamente con una técnica estadística. Comienza con una pregunta que pueda discutirse mediante evidencia.

**Ejemplo:** ¿los programas virtuales tienen la misma presencia en las instituciones oficiales y privadas?

Para responderla debemos precisar:

- **unidad de análisis:** cada programa académico;
- **variables:** modalidad, sector, estado y lugar de oferta;
- **población o muestra:** registros cubiertos por la fuente;
- **método:** limpiar, resumir, comparar e interpretar;
- **límites:** cobertura, valores faltantes, definiciones y fecha de consulta.

Una ruta frecuente es: **pregunta → datos → preparación → análisis → resultado → limitaciones**.


### Dos puntos de partida válidos

1. **Partir de una pregunta.** Se define qué evidencia hace falta y luego se buscan los datos apropiados.
2. **Partir de una base disponible.** Se revisa su contenido, aparecen patrones o anomalías y se formulan preguntas que después deben estudiarse con rigor.

Explorar puede sugerir preguntas; un patrón visual por sí solo todavía no es una conclusión generalizable.


## 2. Investigación cualitativa y cuantitativa

- La investigación **cualitativa** estudia significados, discursos, experiencias y procesos. Python puede apoyar la organización de textos, aunque la interpretación sigue siendo central.
- La investigación **cuantitativa** representa características mediante variables y números para describir, comparar, relacionar o predecir.

Este curso se concentra en datos cuantitativos estructurados. La posibilidad de generalizar no la produce Python: depende del diseño del estudio, la medición y la selección de la población o muestra.


### Propósitos del análisis

| Propósito | Pregunta típica | Resultado |
|---|---|---|
| Descriptivo | ¿Qué está ocurriendo? | porcentajes, tablas, gráficos y medidas |
| Predictivo | ¿Qué valor o categoría podría observarse? | estimación con error medido |
| Prescriptivo | ¿Qué acción conviene bajo ciertos criterios? | alternativas sujetas a restricciones |

Un modelo predictivo no decide por sí mismo. El análisis prescriptivo combina resultados, costos, riesgos, restricciones y criterio humano.

```python
programas_oficiales = 10568
programas_privados = 20593
total_programas = programas_oficiales + programas_privados
proporcion_oficial = programas_oficiales / total_programas
porcentaje_oficial = proporcion_oficial * 100

print("Total:", total_programas)
print("Proporción oficial:", proporcion_oficial)
print("Porcentaje oficial:", porcentaje_oficial)
```

```python
numero = 17
print("Potencia:", numero ** 2)
print("División entera:", numero // 5)
print("Residuo:", numero % 5)

consultas = 2
consultas += 1
print("Asignación abreviada:", consultas)
```

## 8. Comparaciones y valores booleanos

Una comparación produce `True` o `False`. `==` compara igualdad, `!=` diferencia, y `>`, `<`, `>=` y `<=` comparan magnitudes. Recuerda: `=` asigna; `==` compara.


```python
estado_programa = "Activo"
numero_creditos = 144

print(estado_programa == "Activo")
print(estado_programa != "Inactivo")
print(numero_creditos >= 120)
print(numero_creditos < 100)
```

`and` exige que todas las condiciones sean verdaderas. `or` exige al menos una. `not` invierte el valor. En el notebook 2 estos resultados controlarán qué instrucciones se ejecutan.


```python
es_activo = estado_programa == "Activo"
tiene_creditos_suficientes = numero_creditos >= 120

print(es_activo and tiene_creditos_suficientes)
print(es_activo or tiene_creditos_suficientes)
print(not es_activo)
```

### Posiciones y fragmentos

Las posiciones comienzan en cero. `texto[0]` obtiene el primer carácter y `texto[-1]` el último. Un corte `texto[inicio:fin]` incluye el inicio y excluye el final.


```python
palabra = "Investigación"
print(palabra[0])
print(palabra[-1])
print(palabra[0:6])
print(palabra[:6])
print(palabra[6:])
```

### Métodos de texto

Un método se escribe después del valor y un punto. `strip()` quita espacios extremos; `lower()` y `upper()` cambian mayúsculas; `title()` usa mayúscula inicial; `replace()` sustituye fragmentos; `startswith()` y `endswith()` verifican extremos.


```python
nombre_sucio = "  ADMINISTRACIÓN DE EMPRESAS  "
print(nombre_sucio.strip())
print(nombre_sucio.lower())
print(nombre_sucio.strip().title())
print(nombre_sucio.replace("EMPRESAS", "NEGOCIOS"))
print(nombre_sucio.strip().startswith("ADMIN"))
```

Los métodos pueden encadenarse porque cada uno devuelve un texto nuevo. Las cadenas no se modifican en el sitio: debemos asignar el resultado.


```python
municipio_original = "  tunJA  "
municipio_limpio = municipio_original.strip().title()
print("Original:", municipio_original)
print("Limpio:", municipio_limpio)
```

## 10. Listas: varios valores en un orden

Una lista se escribe entre corchetes. Conviene que sus elementos representen la misma clase de información.


```python
paises = ["Colombia", "México", "Perú", "Chile"]
tasas_desempleo = [10.2, 2.8, 5.4, 8.7]
print(paises)
print(tasas_desempleo)
print(type(paises))
print("Cantidad:", len(paises))
```

Las listas usan índices y cortes. El último elemento no tiene siempre la misma posición positiva; `-1` es útil.


```python
print(paises[0])
print(paises[-1])
print(paises[1:3])
print(paises[:2])
```

Una lista es mutable. `append()` agrega al final, `insert()` agrega en una posición, la asignación modifica y `pop()` retira y devuelve un elemento.


```python
indicadores = ["PIB", "Desempleo", "Internet"]
indicadores.append("Inflación")
indicadores.insert(1, "Población")
indicadores[0] = "PIB per cápita"
print(indicadores)

retirado = indicadores.pop()
print("Retirado:", retirado)
print("Final:", indicadores)
```

`sort()` modifica la lista. `sorted(lista)` crea una nueva. Para preservar el orden original de una fuente suele ser mejor crear una copia.


```python
paises_ordenados = sorted(paises)
print("Original:", paises)
print("Ordenada:", paises_ordenados)
print("¿Está Perú?", "Perú" in paises)
```

### Otros métodos y funciones útiles para listas

- `extend()` agrega varios elementos;
- `remove(valor)` elimina la primera coincidencia;
- `count(valor)` cuenta cuántas veces aparece;
- `index(valor)` devuelve la primera posición;
- `copy()` crea otra lista para modificarla sin afectar la original;
- `sum()`, `min()` y `max()` resumen listas numéricas.

Conviene verificar que un valor exista antes de usar `remove()` o `index()`.


```python
indicadores_base = ["PIB", "Desempleo", "Internet"]
indicadores_ampliados = indicadores_base.copy()
indicadores_ampliados.extend(["Inflación", "Población"])

print("Base:", indicadores_base)
print("Ampliada:", indicadores_ampliados)
print("Posición de Internet:", indicadores_ampliados.index("Internet"))
print("Veces que aparece PIB:", indicadores_ampliados.count("PIB"))

if "Inflación" in indicadores_ampliados:
    indicadores_ampliados.remove("Inflación")
print("Después de remove:", indicadores_ampliados)
```

```python
valores = [73.0, 78.6, 74.7, 90.2]
print("Suma:", sum(valores))
print("Mínimo:", min(valores))
print("Máximo:", max(valores))
```

La copia es importante. Si escribimos `otra_lista = lista_original`, ambos nombres apuntan a la misma lista y una modificación se refleja en los dos.


```python
lista_original = ["Colombia", "México"]
mismo_objeto = lista_original
copia_independiente = lista_original.copy()

mismo_objeto.append("Perú")
copia_independiente.append("Chile")

print("Original:", lista_original)
print("Mismo objeto:", mismo_objeto)
print("Copia:", copia_independiente)
```

## 11. Tuplas y conjuntos

Una **tupla** se parece a una lista, pero no se modifica. Un **conjunto** (`set`) conserva valores únicos y no mantiene una posición estable.


```python
coordenada_tunja = (5.5353, -73.3678)
modalidades = {"Presencial", "Virtual", "Presencial", "A distancia"}
print(coordenada_tunja)
print(modalidades)
print("Modalidades únicas:", len(modalidades))
```

Una tupla permite desempaquetar valores. Sus métodos principales son `count()` e `index()`. Si necesitamos modificarla, normalmente era más apropiada una lista.


```python
latitud, longitud = coordenada_tunja
periodos = (2023, 2024, 2024, 2025)

print("Latitud:", latitud)
print("Longitud:", longitud)
print("Veces que aparece 2024:", periodos.count(2024))
print("Posición de 2025:", periodos.index(2025))
```

En conjuntos, `add()` agrega un valor; `update()` agrega varios; `discard()` retira si existe sin generar error. `remove()` también retira, pero falla si el valor no está.


```python
sectores = {"Oficial", "Privado"}
sectores.add("Régimen especial")
sectores.update(["Oficial", "Mixto"])
sectores.discard("Mixto")

print(sectores)
print("¿Oficial está presente?", "Oficial" in sectores)
```

```python
modalidades_2025 = {"Presencial", "Virtual", "A distancia"}
modalidades_2026 = {"Presencial", "Virtual", "Dual"}
print("Unión:", modalidades_2025 | modalidades_2026)
print("En ambos:", modalidades_2025 & modalidades_2026)
print("Solo 2026:", modalidades_2026 - modalidades_2025)
```

## 12. Diccionarios: valores identificados por clave

Un diccionario guarda pares `clave: valor`. Las claves describen cada dato, por lo que es adecuado para una observación individual.

### Crear, consultar y medir un diccionario

Las claves no se consultan por posición sino por nombre. `len()` cuenta cuántas parejas contiene el diccionario.


```python
indicador_internet = {
    "codigo": "IT.NET.USER.ZS",
    "nombre": "Personas que usan Internet",
    "unidad": "% de la población",
    "fuente": "Banco Mundial",
}
print(indicador_internet)
print("Cantidad de claves:", len(indicador_internet))
print(indicador_internet["nombre"])
print(indicador_internet["unidad"])
```

### Agregar, modificar y consultar con seguridad

Podemos agregar o modificar claves con corchetes. `get()` consulta una clave y permite un valor alternativo si no existe.


```python
indicador_internet["anio"] = 2023
indicador_internet["unidad"] = "porcentaje de la población"
print(indicador_internet.keys())
print(indicador_internet.values())
print(indicador_internet.get("periodicidad", "No especificada"))
```

### Vistas, actualización, copia y retiro

`keys()`, `values()` e `items()` ofrecen vistas de claves, valores o parejas. `update()` incorpora varias parejas. `copy()` crea otro diccionario para cambios independientes en este ejemplo sencillo. `pop(clave)` retira una clave y devuelve su valor. El operador `in` comprueba claves, no valores.


```python
indicador_copia = indicador_internet.copy()
indicador_copia.update({
    "periodicidad": "Anual",
    "cobertura": "País"
})

print("Pares clave-valor:", indicador_copia.items())

cobertura_retirada = indicador_copia.pop("cobertura")
print("Retirado:", cobertura_retirada)
print("¿Existe fuente?", "fuente" in indicador_copia)
```

### Diccionarios anidados

Un diccionario puede contener listas u otros diccionarios. La estructura es útil, pero demasiados niveles dificultan leer y validar; para muchas observaciones será más cómodo un DataFrame.


```python
estudio = {
    "titulo": "Conectividad en América Latina",
    "variables": ["internet", "pib_per_capita"],
    "fuente": {
        "nombre": "Banco Mundial",
        "periodicidad": "Anual"
    }
}

print(estudio["variables"][0])
print(estudio["fuente"]["nombre"])
```

### Lista de diccionarios: puente hacia una tabla

Una lista de diccionarios se parece a una tabla: cada diccionario es una fila y cada clave una columna. Pandas convertirá esta estructura en un `DataFrame`.


```python
observaciones = [
    {"pais": "Colombia", "internet": 73.0, "anio": 2023},
    {"pais": "México", "internet": 78.6, "anio": 2023},
    {"pais": "Chile", "internet": 90.2, "anio": 2023},
]
print(observaciones)
print(observaciones[0])
print(observaciones[0]["pais"])
```

### ¿Qué colección conviene elegir?

| Necesidad | Estructura habitual |
|---|---|
| Secuencia ordenada que puede cambiar | lista |
| Secuencia fija | tupla |
| Valores únicos y operaciones de conjuntos | conjunto |
| Una observación con campos identificados | diccionario |
| Muchas observaciones con los mismos campos | lista de diccionarios; después, DataFrame |

La elección depende de cómo se consultarán y modificarán los datos, no de memorizar una estructura “mejor” para todo.
